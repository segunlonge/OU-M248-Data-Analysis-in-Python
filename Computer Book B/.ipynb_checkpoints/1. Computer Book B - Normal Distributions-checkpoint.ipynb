### The proportion of Scottish soldiers who, according to this model, had chests measuring between 37 and 42 inches inclusive.

Essentially the computation in Python is described by the formula below which is a way of standadising data which comes from a normal distribution (or a normally distributed population). The standardisation not only allows for comparisons across different values of a random variable $X$ but also allows for the calculation of probabilities.

The $z$ score represents the so called standard deviation the observed value is from the mean.

$$z = \frac{x - \mu}{\sigma}$$

The lower and upper $z$ scores should be computed and the differences found. Since the area on the normal probability distribution equals 1 the difference between the upper and lower $z$ gives the proportion of Scottish soldiers. The reason the upper and lower $z$ values can be substracted in such a way is that the normal probability curve is a Cummulitative Distribution Curve c.d.f.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Given parameters
mu = 40        # Mean chest size (in inches)
sigma = 2      # Standard deviation (in inches)
lower = 37     # Lower bound
upper = 42     # Upper bound

# Step 1: Calculate the proportion using the Cumulative Distribution Function (CDF) - Z-score calculation
p_lower = stats.norm.cdf(lower, loc=mu, scale=sigma)
p_upper = stats.norm.cdf(upper, loc=mu, scale=sigma)
proportion = p_upper - p_lower

print(f"Proportion between {lower} and {upper} inches: {proportion:.4f} ({proportion * 100:.2f}%)")

# Step 2: Plot the probability distribution
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f'Normal Distribution\n$\\mu={mu}, \\sigma={sigma}$', color='blue', linewidth=2)

# Shade the region between 37 and 42 inches
x_shade = np.linspace(lower, upper, 500)
y_shade = stats.norm.pdf(x_shade, loc=mu, scale=sigma)
plt.fill_between(x_shade, y_shade, color='skyblue', alpha=0.6, label=f'Area = {proportion:.4f} ({proportion * 100:.2f}%)')

# Add labels and styling
plt.title('Chest Measurements of Scottish Soldiers', fontsize=14)
plt.xlabel('Chest Measurement (inches)', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.axvline(mu, color='red', linestyle='--', label=f'Mean ($\\mu = {mu}$)')
plt.legend(loc='upper right')
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
```

#### Commentary

The normal distribution model shows that the proportion of Scottish soldiers that will have chest size between 37 and 42 inches i.e. $P(37 \le X \le 42)$ is roughly around 77.45% i.e. that should be the proportion in a given population.

### The proportion of Scottish soldiers who had chest measurements greater than 43 inches

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Given parameters
mu = 40        # Mean chest size (in inches)
sigma = 2      # Standard deviation (in inches)
threshold = 43 # Chest size threshold

# Step 1: Calculate the proportion using the Survival Function (1 - CDF)
proportion = stats.norm.sf(threshold, loc=mu, scale=sigma)
# Alternatively: proportion = 1 - stats.norm.cdf(threshold, loc=mu, scale=sigma)

print(f"Proportion greater than {threshold} inches: {proportion:.4f} ({proportion * 100:.2f}%)")

# Step 2: Plot the probability distribution
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f'Normal Distribution\n$\\mu={mu}, \\sigma={sigma}$', color='blue', linewidth=2)

# Shade the region greater than 43 inches
x_shade = np.linspace(threshold, mu + 4 * sigma, 500)
y_shade = stats.norm.pdf(x_shade, loc=mu, scale=sigma)
plt.fill_between(x_shade, y_shade, color='orange', alpha=0.6, label=f'Area (X > {threshold}) = {proportion:.4f} ({proportion * 100:.2f}%)')

# Add labels and styling
plt.title('Scottish Soldiers with Chest Size > 43 Inches', fontsize=14)
plt.xlabel('Chest Measurement (inches)', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.axvline(mu, color='red', linestyle='--', label=f'Mean ($\\mu = {mu}$)')
plt.axvline(threshold, color='green', linestyle=':', label=f'Threshold ($X = {threshold}$)')
plt.legend(loc='upper right')
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
```

#### Commentary

The normal distribution model shows that the proportion of Scottish soldiers that will have chest size greater than 37 inches i.e. $P(Z > 1.5) = 1 - 0.9332 = 0.0668$ is around 6.68% which is 1.5 standard deviations (the $z$ score) from the mean

### Below what value were the chest measurements of only 2.5% of Scottish soldiers?

This is a case of doing the reverse of questions a) and b). Questions a) and b) calculations gave the percentages of the observable data by looking up the calculated $z$ score. Now by looking up the percentages in the probabilities for standard normal distribution table to retrieve the $z$ score it is possible to use and re-arrange the $z$ score equation above to calculate the actual observed value $x$ given the constraints of the normal distribution with mean 40 and standard deviation 2.

The re-arrange $z$ score equation then becomes: 

$x = \mu + z \cdot \sigma$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Given parameters
mu = 40          # Mean chest size (in inches)
sigma = 2        # Standard deviation (in inches)
percentile = 0.025 # 2.5%

# Step 1: Calculate the chest measurement using the Percent Point Function (PPF / inverse CDF)
cutoff_value = stats.norm.ppf(percentile, loc=mu, scale=sigma)

print(f"Chest measurement for the lowest {percentile * 100}%: {cutoff_value:.2f} inches")

# Step 2: Plot the probability distribution
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f'Normal Distribution\n$\\mu={mu}, \\sigma={sigma}$', color='blue', linewidth=2)

# Shade the lower 2.5% tail
x_shade = np.linspace(mu - 4 * sigma, cutoff_value, 500)
y_shade = stats.norm.pdf(x_shade, loc=mu, scale=sigma)
plt.fill_between(x_shade, y_shade, color='crimson', alpha=0.6, label=f'Lower 2.5% Area\n($X < {cutoff_value:.2f}$ inches)')

# Add labels and styling
plt.title('Lowest 2.5% of Chest Measurements', fontsize=14)
plt.xlabel('Chest Measurement (inches)', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.axvline(mu, color='black', linestyle='--', label=f'Mean ($\\mu = {mu}$)')
plt.axvline(cutoff_value, color='crimson', linestyle=':', label=f'Cutoff = {cutoff_value:.2f} inches')
plt.legend(loc='upper right')
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
```

#### Commentary

For $P(X < x) = 0.025$ we are interested in the observable value $X$ such that the chest size of Scottish soldiers with value $X$ is less than 2.5% of the all Scottish.

The analysis shows that only 2.5% of Scottish soliders had chest sizes that were below 36.08 inches.

### Chest measurement exceeded by only 5% of Scottish soldiers

Following the same reversal procedure as in question c) we are looking for the value of $X$ of chest measurement that is **exceeded** (exceeded being the key word) by just 5% of Scottish soldiers i.e. $P(X > x) = 0.05$ which is the same as saying  $P(X \le x) = 0.95$ i.e. the value of $X$ which is upto and less that 95%.

The value at the key point where the chest measurement is exceeded by 5% of Scottish soldiers lies at $P(X \le x) = 0.95$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Given parameters
mu = 40          # Mean chest size (in inches)
sigma = 2        # Standard deviation (in inches)
percentile = 0.95 # 95th percentile (exceeded by only 5%)

# Step 1: Calculate the cutoff value using PPF (or inverse Survival Function)
cutoff_value = stats.norm.ppf(percentile, loc=mu, scale=sigma)
# Alternatively: cutoff_value = stats.norm.isf(0.05, loc=mu, scale=sigma)

print(f"Chest measurement exceeded by only 5%: {cutoff_value:.2f} inches")

# Step 2: Plot the probability distribution
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f'Normal Distribution\n$\\mu={mu}, \\sigma={sigma}$', color='blue', linewidth=2)

# Shade the top 5% tail
x_shade = np.linspace(cutoff_value, mu + 4 * sigma, 500)
y_shade = stats.norm.pdf(x_shade, loc=mu, scale=sigma)
plt.fill_between(x_shade, y_shade, color='forestgreen', alpha=0.6, label=f'Top 5% Area\n($X > {cutoff_value:.2f}$ inches)')

# Add labels and styling
plt.title('Chest Measurements Exceeded by Only 5% of Soldiers', fontsize=14)
plt.xlabel('Chest Measurement (inches)', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.axvline(mu, color='black', linestyle='--', label=f'Mean ($\\mu = {mu}$)')
plt.axvline(cutoff_value, color='forestgreen', linestyle=':', label=f'Cutoff = {cutoff_value:.2f} inches')
plt.legend(loc='upper right')
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
```

#### Commentary

The analysis shows that the chest measurement at 43.29 inches is the point at which 5% of Scottish soldiers exceed this chest size.

### a) The probability that an error is positive and greater than 0.5 minutes of arc

The probability can be re-expressed as follows and by calculating the $z$ score with the above formula and taking the compliment we want the shaded region after the value of the random variable $X$ greater than 0.5.

$P(X > 0.5) = P(Z > 0.3015) = 1 - P(Z \le 0.3015)$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# 1. Define distribution parameters
mu = 0               # Mean
variance = 2.75      # Variance
sigma = np.sqrt(variance)  # Standard deviation
threshold = 0.5      # Measurement threshold (minutes of arc)

# 2. Calculate the probability P(X > 0.5)
# 1 - CDF(threshold) gives the area to the right
prob = 1 - stats.norm.cdf(threshold, loc=mu, scale=sigma)
print(f"Probability P(X > {threshold}) = {prob:.4f} (or {prob * 100:.2f}%)")

# 3. Create data for the bell curve (from -4σ to +4σ)
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

# 4. Plot the normal distribution curve
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f"Normal Curve ($\mu = {mu}$, $\sigma^2 = {variance}$)", color="navy", lw=2)

# 5. Shade the region where X > 0.5
x_fill = np.linspace(threshold, mu + 4 * sigma, 500)
y_fill = stats.norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(x_fill, y_fill, color="orange", alpha=0.5, label=f"P(X > {threshold}) = {prob:.4f}")

# 6. Add reference lines
plt.axvline(threshold, color="darkorange", linestyle="--", lw=1.5, label=f"X = {threshold}")
plt.axvline(mu, color="gray", linestyle=":", lw=1.2, label=f"Mean ($\mu = {mu}$)")

# 7. Labels and styling
plt.title("Normal Distribution: Error in Minutes of Arc", fontsize=14)
plt.xlabel("Minutes of Arc", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend(loc="upper right")
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

#### Commentary

The probability that the size of the error (minute of arc) is greater than 0.5 of the observed values is 38.15%

### b) The probability that an error is positive but less than two minutes of arc

Again the question can be re-expressed as:

$P(0 < X < 2) = P(0 < Z < 1.2060) = P(Z < 1.2060) - P(Z < 0)$

Which means considering $z$ score at 0 and and 2 of the random variable $X$. The difference between these two values gives the area on the curve which gives the required probability that the error is positive i.e. greater than zero and less than two minutes of arc.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# 1. Define distribution parameters
mu = 0                    # Mean
variance = 2.75           # Variance
sigma = np.sqrt(variance) # Standard deviation

# Boundaries
lower_bound = 0
upper_bound = 2

# 2. Calculate the probability P(0 < X < 2)
p_lower = stats.norm.cdf(lower_bound, loc=mu, scale=sigma)
p_upper = stats.norm.cdf(upper_bound, loc=mu, scale=sigma)
prob = p_upper - p_lower

print(f"Probability P({lower_bound} < X < {upper_bound}) = {prob:.4f} (or {prob * 100:.2f}%)")

# 3. Create x values for the bell curve (from -4σ to +4σ)
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

# 4. Plot the normal distribution curve
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f"Normal Curve ($\mu = {mu}$, $\sigma^2 = {variance}$)", color="navy", lw=2)

# 5. Shade the region where 0 < X < 2
x_fill = np.linspace(lower_bound, upper_bound, 300)
y_fill = stats.norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(x_fill, y_fill, color="forestgreen", alpha=0.5, 
                 label=f"P({lower_bound} < X < {upper_bound}) = {prob:.4f}")

# 6. Add vertical lines for boundaries
plt.axvline(lower_bound, color="black", linestyle="--", lw=1.2, label=f"X = {lower_bound} (Mean)")
plt.axvline(upper_bound, color="darkgreen", linestyle="--", lw=1.2, label=f"X = {upper_bound}")

# 7. Labels, title, and legend
plt.title("Normal Distribution: $P(0 < X < 2)$ Minutes of Arc", fontsize=14)
plt.xlabel("Minutes of Arc", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend(loc="upper right")
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

#### Commentary

The probability that the size of the error (minute of arc) is between 0 and 2 of the observed values is 38.61%

### c) The probability that the size of an error, which may be positive or negative, is less than 1 minute of arc

So the re-framing is not that the size of an error is less than 1 minute of arc but that it is less than postive 1 minute of arc and greater that negative 1 minute of arc. Hence, the $z$ score at -1 and 1 need to be substracted from each other to find the relevant region under the curve.

$P(|X| < 1) = P(-1 < X < 1) = P(-0.6030 < Z < 0.6030) = P(Z < 0.6030) - P(Z < -0.6030)$

Also, since the normal distribution is perfectly symmetrical the area before -1 and the area after 1 are both the same so we can just express the required area between -1 and 1 as follows:

$P(-0.6030 < Z < 0.6030) = 2 \cdot P(Z < 0.6030) - 1$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# 1. Define distribution parameters
mu = 0                    # Mean
variance = 2.75           # Variance
sigma = np.sqrt(variance) # Standard deviation

# Absolute error threshold: |X| < 1 => -1 < X < 1
lower_bound = -1.0
upper_bound = 1.0

# 2. Calculate the probability P(-1 < X < 1)
p_lower = stats.norm.cdf(lower_bound, loc=mu, scale=sigma)
p_upper = stats.norm.cdf(upper_bound, loc=mu, scale=sigma)
prob = p_upper - p_lower

print(f"Probability P(|X| < 1) = P({lower_bound} < X < {upper_bound}) = {prob:.4f} (or {prob * 100:.2f}%)")

# 3. Create x values for the bell curve (from -4σ to +4σ)
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

# 4. Plot the normal distribution curve
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f"Normal Curve ($\mu = {mu}$, $\sigma^2 = {variance}$)", color="navy", lw=2)

# 5. Shade the region where -1 < X < 1
x_fill = np.linspace(lower_bound, upper_bound, 400)
y_fill = stats.norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(x_fill, y_fill, color="crimson", alpha=0.4, 
                 label=f"P(|X| < 1) = {prob:.4f}")

# 6. Add vertical lines for the boundaries and mean
plt.axvline(lower_bound, color="darkred", linestyle="--", lw=1.2, label=f"X = {lower_bound}")
plt.axvline(upper_bound, color="darkred", linestyle="--", lw=1.2, label=f"X = {upper_bound}")
plt.axvline(mu, color="gray", linestyle=":", lw=1.2, label=f"Mean ($\mu = {mu}$)")

# 7. Labels, title, and styling
plt.title("Normal Distribution: Error Size Less Than 1 Minute of Arc ($|X| < 1$)", fontsize=14)
plt.xlabel("Minutes of Arc", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend(loc="upper right")
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

#### Commentary

The probability that the size of the error of observed values is less than -1 (in absolute terms) and less than 1 lies between -1 and 1 and is 45.35%

### d) What is the probability that the size of an error, positive or negative, is greater than 1.5 minutes of arc?

Re-expressing this means to find the required probability that the error (minute of arc) is greater than positive 1.5 and negative 1.5 so:

$P(|X| > 1.5) = P(X < -1.5) + P(X > 1.5)$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# 1. Distribution parameters
mu = 0
variance = 2.75
sigma = np.sqrt(variance)
threshold = 1.5

# 2. Probability calculation: P(X < -1.5) + P(X > 1.5)
p_left = stats.norm.cdf(-threshold, loc=mu, scale=sigma)
p_right = 1 - stats.norm.cdf(threshold, loc=mu, scale=sigma)
prob = p_left + p_right

print(f"Probability P(|X| > {threshold}) = {prob:.4f} (or {prob * 100:.2f}%)")

# 3. Plot bell curve
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f"Normal Curve ($\mu = {mu}$, $\sigma^2 = {variance}$)", color="navy", lw=2)

# 4. Shade both tails (|X| > 1.5)
# Left tail
x_left = np.linspace(mu - 4 * sigma, -threshold, 300)
y_left = stats.norm.pdf(x_left, loc=mu, scale=sigma)
plt.fill_between(x_left, y_left, color="purple", alpha=0.5, label=f"P(|X| > {threshold}) = {prob:.4f}")

# Right tail
x_right = np.linspace(threshold, mu + 4 * sigma, 300)
y_right = stats.norm.pdf(x_right, loc=mu, scale=sigma)
plt.fill_between(x_right, y_right, color="purple", alpha=0.5)

# 5. Add boundary lines
plt.axvline(-threshold, color="darkviolet", linestyle="--", lw=1.2, label=f"X = -{threshold}")
plt.axvline(threshold, color="darkviolet", linestyle="--", lw=1.2, label=f"X = +{threshold}")
plt.axvline(mu, color="gray", linestyle=":", lw=1.2, label=f"Mean ($\mu = {mu}$)")

# 6. Styling
plt.title("Normal Distribution: Error Size Greater Than 1.5 Minutes of Arc ($|X| > 1.5$)", fontsize=13)
plt.xlabel("Minutes of Arc", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend(loc="upper right")
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

#### Commentary

Remembering that the area under the normal distribution curve is also equal to one then the area before the value of -1.5 is greater than -1.5 in abosulte terms (not in relative terms) and of course the area after the value of 1.5 is the probability required. These two areas are symmetrical and identical.

Therefore, the probability that the observed value is before -1.5 and greater than 1.5 is 36.57%.

### e) Suppose that 10% of errors are positive and greater than $x$. Find the value of $x$.

The question to answer here is given the known mean of: $\mu$ = $0$ and standard deviation of: $\sigma$ $\approx$ $\sqrt{2.75} \approx 1.6583$ what should the observed value of errors $X$ be given that probability that it is 10% and positive


Re-arranging the $z$ score equation above results in: $x = \mu + z\sigma$. The question is the reverse of $P(X > x) = 0.10$ i.e. what is the value of $x$ such that 10% of errors are positive and are greater than $x$.

To the right of $x$ are where 10% of errors lay which are positive but given the normal distribution curve is a Cumulative Distribution Function (CDF) the value of $x$ is less than and equal at $x$ is what is required i.e. $P(X \le x) = 1 - 0.10 = 0.90$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# 1. Distribution parameters
mu = 0
variance = 2.75
sigma = np.sqrt(variance)

# 2. Find x such that P(X > x) = 0.10
# Using percent point function (ppf / inverse CDF) for 90th percentile
prob_left = 0.90
z_score = stats.norm.ppf(prob_left)
x_val = stats.norm.ppf(prob_left, loc=mu, scale=sigma)

print(f"z-score: {z_score:.4f}")
print(f"Value of x: {x_val:.4f} minutes of arc")

# 3. Plot bell curve
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f"Normal Curve ($\mu = {mu}$, $\sigma^2 = {variance}$)", color="navy", lw=2)

# 4. Shade top 10% region (X > x)
x_fill = np.linspace(x_val, mu + 4 * sigma, 300)
y_fill = stats.norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(x_fill, y_fill, color="tomato", alpha=0.5, label=f"Top 10% (Area = 0.10)")

# 5. Add reference lines
plt.axvline(x_val, color="red", linestyle="--", lw=1.5, label=f"x = {x_val:.2f}")
plt.axvline(mu, color="gray", linestyle=":", lw=1.2, label=f"Mean ($\mu = {mu}$)")

# 6. Styling
plt.title("Normal Distribution: 90th Percentile ($P(X > x) = 0.10$)", fontsize=13)
plt.xlabel("Minutes of Arc", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend(loc="upper right")
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

#### Commentary

The value of $x$ such that 10% of errors are positive and are **greater** than $x$ is 2.13


### f) Suppose that 95% of errors lie between $−b$ and $b$ (where $b$ is positive). Find the value of $b$.

The question should be re-phrased as follows: $P(-b < X < b) = 0.95$ i.e. the values of $b$ in the ranges such that the random variable has a 95% occurance within that range. Remembering that the area under the cover is 1 or 100% then it follows that the area between $-b$ and $b$ is of course 95% but that from left to $b$ the area has to be 0.95 + 0.025 (since on either side of $-b$ and $b$ the area is 0.025 i.e. 5% divided by 2 due to symmetry). This value in the $z$ lookup table has a $z$ value of: 1.960. By reversing the process of finding the probability from the random variable $X$ we can now find the value of $b$ ($X$) from the probaility of 0.975 by re-arranging the equation:

$$z = \frac{X - \mu}{\sigma}$$ as 

$$x = \mu + z\sigma$$

Given:

$\mu$ = $0$<br>
$\sigma$: $\sqrt{2.75} \approx 1.6583$<br>
$z = 1.960$

Then:

$x = 0 + (1.960 \times 1.6583)$ $\approx$ $b \approx 3.2503$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# 1. Distribution parameters
mu = 0
variance = 2.75
sigma = np.sqrt(variance)

# 2. Find b for central 95% (tail = 2.5%, upper percentile = 97.5%)
prob_central = 0.95
alpha = 1 - prob_central
prob_upper = 1 - (alpha / 2)  # 0.975

z_val = stats.norm.ppf(prob_upper)
b_val = stats.norm.ppf(prob_upper, loc=mu, scale=sigma)

print(f"z-score: {z_val:.4f}")
print(f"Value of b: {b_val:.4f} minutes of arc")

# 3. Plot bell curve
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f"Normal Curve ($\mu = {mu}$, $\sigma^2 = {variance}$)", color="navy", lw=2)

# 4. Shade central 95% region (-b to b)
x_fill = np.linspace(-b_val, b_val, 500)
y_fill = stats.norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(x_fill, y_fill, color="teal", alpha=0.45, label=f"Central 95% (Area = 0.95)")

# 5. Add boundary lines
plt.axvline(-b_val, color="darkcyan", linestyle="--", lw=1.5, label=f"-b = -{b_val:.2f}")
plt.axvline(b_val, color="darkcyan", linestyle="--", lw=1.5, label=f"+b = +{b_val:.2f}")
plt.axvline(mu, color="gray", linestyle=":", lw=1.2, label=f"Mean ($\mu = {mu}$)")

# 6. Labels and styling
plt.title("Normal Distribution: Central 95% Interval ($-b < X < b$)", fontsize=13)
plt.xlabel("Minutes of Arc", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend(loc="upper right")
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

#### Commentary

The value of $b$ at which 95% of the errors lay (between $-b$ and $b$) was found to be approximately 3.2502 minutes of arc by taking account of the symmetry on either side of $-b$ and $b$ and the area from the left upto and including $b$

## Computer activity 3: Heights of elderly women

Suppose that the variation in the heights (in centimetres) of elderly women may be adequately modelled by a normal distribution with mean 160 and standard deviation 6.

(a) According to the model, what proportion of elderly women are taller than 166 cm?<br>
(b) Find the probability that the height of a randomly selected woman will be between 145 cm and 157 cm.<br>
(c) Below what value are the heights of 85% of elderly women?<br>
(d) Find the interquartile range of the heights of the population of elderly women.<br>

### a) According to the model, what proportion of elderly women are taller than 166 cm

Re-phrasing the question: $P(X > 166)$ i.e. the probability of the random variable being greater than 166cm.

The $z$ value i.e. the observed value standadised is: $z = \frac{166 - 160}{6} = \frac{6}{6} = 1$.

Finding the region to the right of 1 on the standard normal distribution: $P(Z > 1) = 1 - P(Z \le 1) = P(Z > 1) = 1 - 0.8413 = 0.1587$

Hence the proportion of elderly women taller than 166cm is 15.87%

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# Parameters
mu = 160  # Mean height in cm
sigma = 6  # Standard deviation in cm
cutoff = 166  # Height cutoff in cm

# Calculate the proportion (probability)
z_score = (cutoff - mu) / sigma
proportion = 1 - stats.norm.cdf(cutoff, loc=mu, scale=sigma)
print(f"z-score: {z_score:.2f}")
print(f"Proportion taller than {cutoff} cm: {proportion:.4f} ({proportion * 100:.2f}%)")

# Generate x-values covering 4 standard deviations on either side of the mean
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

# Create the plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, "b-", linewidth=2, label="Normal Distribution")

# Shade the area where height > 166 cm
x_fill = np.linspace(cutoff, mu + 4 * sigma, 500)
y_fill = stats.norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(
    x_fill,
    y_fill,
    color="orange",
    alpha=0.5,
    label=f"P(X > {cutoff}) ≈ {proportion:.4f}",
)

# Add reference lines for the mean and cutoff
plt.axvline(
    mu,
    color="gray",
    linestyle="--",
    linewidth=1.5,
    label=f"Mean (µ = {mu} cm)",
)
plt.axvline(
    cutoff,
    color="red",
    linestyle="--",
    linewidth=1.5,
    label=f"Cutoff (X = {cutoff} cm)",
)

# Labels and title
plt.title(
    "Height Distribution of Elderly Women\n$N(\\mu = 160, \\sigma^2 = 6^2)$",
    fontsize=14,
)
plt.xlabel("Height (cm)", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)

# Display the plot
plt.tight_layout()
plt.show()
```

#### Commentary

The proportion of elderly women taller than 166cm is 15.87% assumming the height of elderly women is normally distributed.

#### b) Find the probability that the height of a randomly selected woman will be between 145 cm and 157 cm

The question can be re-phrased as follows: $P(145 < X < 157)$.

Hence we need to calculate the $z$ score at two points (145 and 157) and substract the difference to find the area of this region.

$z$ value upto 145:

$z_1 = \frac{145 - 160}{6} = \frac{-15}{6} = -2.5$ $\approx 0.3085$

$z$ value upto 157:

$z_2 = \frac{157 - 160}{6} = \frac{-3}{6} = -0.5$ $\approx 0.0062$

Therefore, substracting $z$ values:

$P(145 < X < 157) = 0.3085 - 0.0062 = 0.3023$

Hence the proportion of elderly women with height between 145 cm and 157 cm is 30.23%

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# Parameters
mu = 160  # Mean height in cm
sigma = 6  # Standard deviation in cm
lower_bound = 145  # Lower limit in cm
upper_bound = 157  # Upper limit in cm

# Calculate z-scores
z1 = (lower_bound - mu) / sigma
z2 = (upper_bound - mu) / sigma

# Calculate the probability between bounds
prob_lower = stats.norm.cdf(lower_bound, loc=mu, scale=sigma)
prob_upper = stats.norm.cdf(upper_bound, loc=mu, scale=sigma)
probability = prob_upper - prob_lower

print(f"z-score for {lower_bound} cm: {z1:.2f}")
print(f"z-score for {upper_bound} cm: {z2:.2f}")
print(
    f"Probability between {lower_bound} cm and {upper_bound} cm: {probability:.4f} ({probability * 100:.2f}%)"
)

# Generate x-values covering 4 standard deviations around the mean
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

# Create the plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, "b-", linewidth=2, label="Normal Distribution")

# Shade the area between 145 cm and 157 cm
x_fill = np.linspace(lower_bound, upper_bound, 500)
y_fill = stats.norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(
    x_fill,
    y_fill,
    color="green",
    alpha=0.5,
    label=f"P({lower_bound} < X < {upper_bound}) ≈ {probability:.4f}",
)

# Reference lines
plt.axvline(
    mu,
    color="gray",
    linestyle="--",
    linewidth=1.5,
    label=f"Mean (µ = {mu} cm)",
)
plt.axvline(
    lower_bound,
    color="red",
    linestyle=":",
    linewidth=1.5,
    label=f"Lower = {lower_bound} cm",
)
plt.axvline(
    upper_bound,
    color="purple",
    linestyle=":",
    linewidth=1.5,
    label=f"Upper = {upper_bound} cm",
)

# Labels, title, and legend
plt.title(
    "Height Distribution of Elderly Women\n$P(145 < X < 157)$", fontsize=14
)
plt.xlabel("Height (cm)", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend(fontsize=10, loc="upper right")
plt.grid(True, alpha=0.3)

# Display the plot
plt.tight_layout()
plt.show()
```

#### Commentary

Hence the proportion of elderly women with height between 145 cm and 157 cm is 30.23% assumming the height of elderly women is normally distributed.

#### c) Below what value are the heights of 85% of elderly women?

In the question we are given a probability value of 85% and asked to find the value of heights such that $x$ is a value in height which accounts for 85% of elderly women. Hence, rephrasing the question as: $P(X < x) = 0.85$.

Essentially, this is about finding the value of $x$ in the $z$ score equation from the random variable $X$:
$$z = \frac{x - \mu}{\sigma}$$

Re-arranging gives:

$$x = \mu + z\sigma$$

Hence plugging in the values: $x = 160 + (1.0364 \times 6)$ $\approx 166.22\text{ cm}$


```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# Parameters
mu = 160  # Mean height in cm
sigma = 6  # Standard deviation in cm
percentile = 0.85  # 85%

# Find the z-score and corresponding height (x-value) using the Percent Point Function (inverse CDF)
z_score = stats.norm.ppf(percentile)
x_cutoff = stats.norm.ppf(percentile, loc=mu, scale=sigma)

print(f"z-score for 85th percentile: {z_score:.4f}")
print(f"Height below which 85% of women lie: {x_cutoff:.2f} cm")

# Generate x-values covering 4 standard deviations on either side of the mean
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

# Create the plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, "b-", linewidth=2, label="Normal Distribution")

# Shade the area below the 85th percentile cutoff
x_fill = np.linspace(mu - 4 * sigma, x_cutoff, 500)
y_fill = stats.norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(
    x_fill,
    y_fill,
    color="skyblue",
    alpha=0.6,
    label=f"P(X < {x_cutoff:.2f} cm) = 85%",
)

# Add reference lines
plt.axvline(
    mu,
    color="gray",
    linestyle="--",
    linewidth=1.5,
    label=f"Mean (µ = {mu} cm)",
)
plt.axvline(
    x_cutoff,
    color="red",
    linestyle="--",
    linewidth=1.5,
    label=f"85th Percentile = {x_cutoff:.2f} cm",
)

# Labels, title, and formatting
plt.title(
    "Height Distribution of Elderly Women\n85th Percentile ($X \\approx 166.22\\text{ cm}$)",
    fontsize=14,
)
plt.xlabel("Height (cm)", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend(fontsize=11, loc="upper right")
plt.grid(True, alpha=0.3)

# Display the plot
plt.tight_layout()
plt.show()
```

#### d) Find the interquartile range of the heights of the population of elderly women.

The interquartile range is simply the middle 50% of a data set i.e. $\text{IQR} = Q_3 - Q_1$. So the question is asking the range of heights between the upper quarter (75%) and the lower quarter (25%) of elderly women that fall within the whole population.

Again this boils down to finding the value to the $z$ score at q3 and q1 to use to find the value of $x$ according to the equation below:
$$x = \mu + z\sigma$$

The value of $z$ in the z-table lookup for 0.25 has a $z$ value of around $\approx -0.6745$ and by symmetry the value of $z$ in the z-table lookup for 0.75 has a $z$ value of around $\approx +0.6745$.

With the given values of $\mu = 160$ and  $\sigma = 6$ these can be plugged into the equation above and the difference of $x_3$ and $x_1$ found as the interquartile range of the heights of the population of elderly women. 

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# Parameters
mu = 160  # Mean height in cm
sigma = 6  # Standard deviation in cm

# Calculate Q1 (25th percentile) and Q3 (75th percentile)
q1 = stats.norm.ppf(0.25, loc=mu, scale=sigma)
q3 = stats.norm.ppf(0.75, loc=mu, scale=sigma)
iqr = q3 - q1

# Z-scores
z1 = stats.norm.ppf(0.25)
z3 = stats.norm.ppf(0.75)

print(f"z-score for Q1: {z1:.4f}")
print(f"z-score for Q3: {z3:.4f}")
print(f"Q1 (25th percentile): {q1:.2f} cm")
print(f"Q3 (75th percentile): {q3:.2f} cm")
print(f"Interquartile Range (IQR = Q3 - Q1): {iqr:.2f} cm")

# Generate x-values covering 4 standard deviations around the mean
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(x, y, "b-", linewidth=2, label="Normal Distribution")

# Shade the IQR region (middle 50% between Q1 and Q3)
x_fill = np.linspace(q1, q3, 500)
y_fill = stats.norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(
    x_fill,
    y_fill,
    color="mediumpurple",
    alpha=0.5,
    label=f"IQR (Middle 50%) = {iqr:.2f} cm",
)

# Reference lines for Mean, Q1, and Q3
plt.axvline(
    mu,
    color="gray",
    linestyle="--",
    linewidth=1.5,
    label=f"Mean (µ = {mu} cm)",
)
plt.axvline(
    q1,
    color="darkorange",
    linestyle="--",
    linewidth=1.5,
    label=f"Q1 = {q1:.2f} cm",
)
plt.axvline(
    q3,
    color="forestgreen",
    linestyle="--",
    linewidth=1.5,
    label=f"Q3 = {q3:.2f} cm",
)

# Annotate the IQR width with a double-headed arrow
arrow_y = stats.norm.pdf(mu, loc=mu, scale=sigma) * 0.5
plt.annotate(
    "",
    xy=(q1, arrow_y),
    xytext=(q3, arrow_y),
    arrowprops=dict(arrowstyle="<->", color="black", lw=1.5),
)
plt.text(
    mu,
    arrow_y * 1.08,
    f"IQR = {iqr:.2f} cm",
    ha="center",
    fontsize=11,
    fontweight="bold",
)

# Labels, title, and styling
plt.title(
    "Height Distribution of Elderly Women\nInterquartile Range ($Q_1$ to $Q_3$)",
    fontsize=14,
)
plt.xlabel("Height (cm)", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend(fontsize=10, loc="upper right")
plt.grid(True, alpha=0.3)

# Display the plot
plt.tight_layout()
plt.show()
```

#### Commentary

The interquartile range of the heights of the population of elderly women is 8.09 cm. It gives a good measure of spread around the mean (because it ignores the top and bottom quarter to avoid distortion) in this case there is not much spread in height amongst elderly women according to the sample with reference to the population.

## Computer activity 4: Nicotine levels of smokers

Suppose that the variation in the blood plasma nicotine levels (in nanograms per millilitre) of smokers may be modelled by a normal
distribution with mean 315 and standard deviation 131.

(a) According to the model, what proportion of smokers have nicotine levels lower than 300?<br>
(b) What proportion of smokers have nicotine levels between 150 and 400?<br>
(c) What nicotine level is such that only 20% of smokers have a higher level?<br>

### a) According to the model, what proportion of smokers have nicotine levels lower than 300?

Re-phrasing the question the population of smokers that have nicotine levels lower than 300 would be:

$P(X < 300)$

And using the $z$ score equation to calculate the value of $z$ as follows:

$z = \frac{300 - 315}{131} \approx = -0.1145$

Hence, 

$P(Z < -0.1145) \approx 0.4544$


```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats

# 1. Define distribution parameters
mu = 315  # Mean nicotine level
sigma = 131  # Standard deviation
cutoff = 300  # Value of interest

# 2. Calculate the proportion P(X < 300)
proportion = stats.norm.cdf(cutoff, loc=mu, scale=sigma)
print(f"Proportion of smokers with nicotine levels < {cutoff}: {proportion:.4f} ({proportion * 100:.2f}%)")

# 3. Generate x values for the normal curve (covering 4 standard deviations)
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

# 4. Create the plot
plt.figure(figsize=(9, 5))
plt.plot(x, y, label=f"Normal Distribution\n($\mu={mu}$, $\sigma={sigma}$)", color="blue", lw=2)

# Shade the region where X < 300
x_fill = np.linspace(mu - 4 * sigma, cutoff, 500)
y_fill = stats.norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(x_fill, y_fill, color="skyblue", alpha=0.5, label=f"$P(X < 300) \\approx {proportion:.4f}$")

# Add reference lines for mean and cutoff value
plt.axvline(mu, color="black", linestyle="--", alpha=0.7, label=f"Mean ($\mu={mu}$)")
plt.axvline(cutoff, color="red", linestyle="--", alpha=0.8, label=f"Cutoff ($X={cutoff}$)")

# Labels and styling
plt.title("Blood Plasma Nicotine Levels in Smokers", fontsize=14, fontweight="bold")
plt.xlabel("Nicotine Level (ng/mL)", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper right")
plt.tight_layout()

# Display the plot
plt.show()
```

#### Commentary

According to the normal distribution model, approximately $0.4544$ (or $45.44\%$) of smokers have nicotine levels lower than $300\text{ ng/mL}$.

### What proportion of smokers have nicotine levels between 150 and 400?

Re-phrasing the question the proportion of smokers that have nicotine levels between 150 and 400 is:

$P(150 < X < 300)$

Then calculating the $z$ scores at both levels and substracting the difference of the lookup values gives the proportion of smokers with nicotine levels between 150 and 400.

Hence:

At $x = 150$<br>
$z_1 = \frac{150 - 315}{131} = \frac{-165}{131} \approx -1.26$

$P(X < 150) = P(Z < -1.26) \approx 0.1039$

At $x = 400$<br>
$z_2 = \frac{400 - 315}{131} = \frac{85}{131} \approx 0.65$

$P(X < 400) = P(Z < 0.65) \approx 0.7418$

Therefore:

$P(150 < X < 400) = 0.7418 - 0.1039 = 0.6379$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters of the normal distribution
mu = 315       # Mean
sigma = 131    # Standard deviation

# Interval
lower = 150
upper = 400

# Calculate the proportion between 150 and 400
proportion = norm.cdf(upper, loc=mu, scale=sigma) - \
             norm.cdf(lower, loc=mu, scale=sigma)

print(f"Proportion: {proportion:.4f}")
print(f"Percentage: {proportion * 100:.2f}%")

# Create x-values and normal probability density values
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1_000)
y = norm.pdf(x, loc=mu, scale=sigma)

# Values used for shading the area between 150 and 400
x_shaded = np.linspace(lower, upper, 500)
y_shaded = norm.pdf(x_shaded, loc=mu, scale=sigma)

# Create the plot
plt.figure(figsize=(10, 6))

# Plot the normal distribution
plt.plot(
    x,
    y,
    color="blue",
    linewidth=2,
    label=fr"Normal distribution ($\mu={mu}$, $\sigma={sigma}$)"
)

# Shade the area between 150 and 400
plt.fill_between(
    x_shaded,
    y_shaded,
    color="skyblue",
    alpha=0.6,
    label=fr"$P({lower} < X < {upper}) \approx {proportion:.4f}$"
)

# Add vertical lines at the boundaries and mean
plt.axvline(lower, color="red", linestyle="--", label=f"Lower bound = {lower}")
plt.axvline(upper, color="green", linestyle="--", label=f"Upper bound = {upper}")
plt.axvline(mu, color="black", linestyle=":", label=f"Mean = {mu}")

# Add labels and title
plt.title("Blood Plasma Nicotine Levels")
plt.xlabel("Nicotine level (ng/mL)")
plt.ylabel("Probability density")

plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()
plt.tight_layout()

# Display the graph
plt.show()
```

#### Commentary

According to the normal distribution model the proportion of smokers with nicotine levels between $150\text{ ng/mL}$ and $400\text{ ng/mL}$ is $63.79\%$ 

### c) What nicotine level is such that only 20% of smokers have a higher level?
